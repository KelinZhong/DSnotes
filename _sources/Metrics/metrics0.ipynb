# Business Metrics

Landing page for the **Metrics** chapter — a paired reference for the business metrics that come up in data science interviews (analytics rounds, metric-definition questions, and SQL/pandas coding rounds).

This chapter contains two notebooks that cover the **same 12 metrics in the same order**, so you can study each metric in both languages side by side:

- **Business Metrics — Python Reference** ({doc}`metrics_Python <metrics_Python>`): computes every metric with pandas on a simulated product dataset (users, events, orders, marketing spend, subscriptions, ad events, marketplace transactions, NPS responses), with plots for the time-series metrics and a pandas cheat sheet at the end.
- **Business Metrics — SQL Reference** ({doc}`metrics_SQL <metrics_SQL>`): the same metrics written as SQL queries against a shared schema, with a metric-formula quick reference at the end.

Because the section numbering (§1–§12) matches across the two notebooks, a good drill is: pick a metric, write the SQL from memory, then write the pandas version, and check both against the reference.

## The 12 metrics covered

| § | Metric | What it measures | Typical interview angle |
|---|--------|------------------|-------------------------|
| 1 | **DAU / WAU / MAU** | Daily / weekly / monthly active users | Define "active"; dedupe users per window |
| 2 | **DAU/MAU ratio** | Stickiness — how often monthly users return | Why 20% is a common benchmark; rolling averages |
| 3 | **D1 / D7 / D30 retention** | Share of a signup cohort active N days later | Cohort tables; retention curves |
| 4 | **K-factor** | Viral growth — invites sent × invite conversion | When K > 1 matters; monthly trend |
| 5 | **NPS** | Net Promoter Score = % promoters − % detractors | Score buckets (0–6 / 7–8 / 9–10) |
| 6 | **Conversion rate** | Visitors → purchasers; multi-step funnel drop-off | Funnel analysis; step-wise vs. overall conversion |
| 7 | **LTV** | Customer lifetime value | Per-customer aggregation; LTV by channel |
| 8 | **CAC** | Customer acquisition cost | Spend ÷ new customers; CAC by channel |
| 9 | **LTV/CAC ratio** | Unit economics health | Why ~3× is a common target |
| 10 | **MRR & growth rate** | Monthly recurring revenue (SaaS) | MRR by fee tier; month-over-month growth |
| 11 | **CTR** | Click-through rate on ads/content | Daily CTR; CTR by ad |
| 12 | **GMV** | Gross merchandise value (marketplace) | GMV vs. revenue distinction |
